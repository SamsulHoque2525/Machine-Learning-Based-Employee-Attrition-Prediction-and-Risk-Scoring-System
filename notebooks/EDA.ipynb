```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/Palo Alto Networks.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# Attrition distribution

attrition_counts = df["Attrition"].value_counts()

print("Attrition Count:")
print(attrition_counts)

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Attrition"
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Attrition by Overtime

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="OverTime",
    hue="Attrition"
)

plt.title("Employee Attrition by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Attrition by Department

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition"
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.xticks(rotation=15)

plt.show()
```

```python
# Attrition by Job Role

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="JobRole",
    hue="Attrition"
)

plt.title("Employee Attrition by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

```python
# Job Satisfaction vs Attrition

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="JobSatisfaction",
    hue="Attrition"
)

plt.title("Job Satisfaction vs Employee Attrition")
plt.xlabel("Job Satisfaction (1 = Low, 4 = High)")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Age distribution by attrition

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Age",
    hue="Attrition",
    bins=20,
    kde=True
)

plt.title("Age Distribution and Employee Attrition")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Monthly Income vs Attrition

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income vs Employee Attrition")
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.ylabel("Monthly Income")

plt.show()
```
